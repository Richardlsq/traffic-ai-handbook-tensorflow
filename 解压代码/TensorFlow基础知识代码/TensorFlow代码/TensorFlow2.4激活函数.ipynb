```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10,10)
y_sigmoid = 1/(1+np.exp(-x))
y_tanh = (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))

fig = plt.figure()
# plot sigmoid
plt.plot(x,y_sigmoid)
plt.grid()
plt.title('Sigmoid')
plt.show()
```

```python
import tensorflow as tf

input_x = tf.constant([-2., -0.5, 0.5, 2.])
manual = 1 / (1 + tf.exp(-input_x))
print(manual)
```

```python
output = tf.sigmoid(input_x)
tf.debugging.assert_near(manual, output)
print("Sigmoid matches:", output)
```

```python
import tensorflow as tf

input_x = tf.constant([-2., -0.5, 0.5, 2.])
manual = tf.where(input_x >= 0, input_x, tf.zeros_like(input_x))
print(manual)
```

```python
output = tf.nn.relu(input_x)
tf.debugging.assert_near(manual, output)
print("ReLU matches:", output)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10,10)
y_sigmoid = 1/(1+np.exp(-x))
y_tanh = (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))

fig = plt.figure()
ax = fig.add_subplot()
y_relu = np.array([0.01*item  if item<0 else item for item in x ])
ax.plot(x,y_relu)
ax.grid()
ax.set_title('Leaky ReLu')
plt.show()
```

```python
import tensorflow as tf

input_x = tf.constant([-2., -0.5, 0.5, 2.])
gamma = 0.01
manual = tf.where(input_x >= 0, input_x, gamma * input_x)
print(manual)
```

```python
output = tf.nn.leaky_relu(input_x, alpha=gamma)
tf.debugging.assert_near(manual, output)
print("Leaky ReLU matches:", output)
```
