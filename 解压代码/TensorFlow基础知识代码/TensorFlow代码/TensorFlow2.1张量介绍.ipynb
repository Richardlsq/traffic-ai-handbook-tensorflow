```python
import tensorflow as tf
import numpy as np

# 从 NumPy 数组和列表构建张量
a = np.array([[1, 2, 3], [4, 5, 6]])
x = tf.convert_to_tensor(a, dtype=tf.float32)
print(a, x)
x = tf.constant([[1, 2]], dtype=tf.float32)
print(x)
```

```python
temp1 = tf.random.stateless_uniform((5,), seed=(1, 0))
temp2 = tf.random.stateless_uniform((5,), seed=(1, 0))
temp3 = tf.random.stateless_uniform((5,), seed=(2, 0))
tf.debugging.assert_equal(temp1, temp2)
print('Same seed repeats:', bool(tf.reduce_all(temp1 == temp2)))
print('Different seed differs:', not bool(tf.reduce_all(temp1 == temp3)))
```

```python
tensor1 = tf.zeros([2, 3])
tensor2 = tf.ones([2, 3])
tensor3 = tf.eye(3)
print(tensor1, '\n', tensor2, '\n', tensor3)
```

```python
# 创建时指定类型，也可以在创建后转换。
x = tf.ones((2, 3, 4), dtype=tf.int64)
print(x)
x = tf.cast(tf.ones((2, 3, 4)), tf.int64)
print(x)
```

```python
x = tf.random.uniform((3, 2))
print(x.shape)
y = tf.reshape(x, (2, 3))
print(y.shape)
```

```python
a = tf.random.uniform((3, 4))
b = tf.expand_dims(a, 0)
c = a[None, ...]
print(b.shape, c.shape)
```

```python
a = tf.random.uniform((1, 1, 3, 4))
b = tf.squeeze(a)
c = tf.squeeze(a, axis=1)
print(b.shape, c.shape)
```

```python
a = tf.random.uniform((1, 3, 28, 32))
# 从 (批次, 通道, 高, 宽) 交换到 (批次, 高, 宽, 通道)。
b = tf.transpose(a, (0, 2, 3, 1))
print(b.shape)
```

```python
# tf.concat()拼接方法的代码如下：
a = tf.random.uniform([1, 2])
b = tf.random.uniform([1, 2])
c = tf.random.uniform([1, 2])
output1 = tf.concat([a, b, c], axis=0)  # axis=0：增加行数
print(output1.shape)
output2 = tf.concat([a, b, c], axis=1)  # axis=1：增加列数
print(output2.shape)
```

```python
# 沿轴按指定长度切分。
a = tf.random.uniform((3, 4))
output1 = tf.split(a, [2, 1], axis=0)
print(output1)
output2 = tf.split(a, [2, 2], axis=1)
print(output2)
```

```python
# tf.stack()堆叠方法的代码如下：
a = tf.random.uniform([1, 2])
b = tf.random.uniform([1, 2])
c = tf.random.uniform([1, 2])
output1 = tf.stack([a, b, c], axis=0)  # 在第 0 轴插入新维度
print(output1.shape)
output2 = tf.stack([a, b, c], axis=1)  # 在第 1 轴插入新维度
print(output2.shape)
```

```python
a = tf.random.uniform((3, 4))
# 整数指定等长分块的块数，长度不等时传入每块大小。
print(tf.split(a, 2, axis=1))
print(tf.split(a, [2, 1], axis=0))
```

```python
x = tf.random.uniform([2, 3, 4])
print(x[1].shape)

y = x[1, 0:2, :]  # 等价于 x[1][0:2]
print(y.shape)

z = x[:, 0, ::2]
print(z.shape)
```

```python
# 元素求和第一种方法
a = tf.random.uniform([4, 3])
b = tf.reduce_sum(a)
print(b)
```

```python
# 元素求和第二种方法
a = tf.random.uniform([4, 3])
b = tf.reduce_sum(a, axis=1, keepdims=True)
print(b, b.shape)
```

```python
# 沿第 0 轴按索引累加。
x = tf.constant([[1., 2.], [3., 4.]])
y = tf.constant([[3., 4.], [5., 6.]])
output = tf.tensor_scatter_nd_add(x, [[0], [1]], y)
print(output)
```

```python
# 元素乘积第一种方法
a = tf.random.uniform([4, 3])
b = tf.reduce_prod(a)
print(b)
```

```python
# 元素乘积第二种方法
a = tf.random.uniform([4, 3])
b = tf.reduce_prod(a, 1, True)
print(b, b.shape)
```

```python
# 求平均数的第一种方法
a = tf.random.uniform([4, 3])
b = tf.reduce_mean(a)
print(b)
```

```python
# 求平均数的第二种方法
a = tf.random.uniform([4, 3])
b = tf.reduce_mean(a, 1, True)
print(b, b.shape)
```

```python
# 全部元素的样本方差（分母为 n − 1）。
a = tf.random.uniform((4, 3))
n = tf.cast(tf.size(a), a.dtype)
b = tf.math.reduce_variance(a) * n / (n - 1)
print(b)
```

```python
# 沿第 1 轴求样本方差，保留该轴。
a = tf.random.uniform((4, 3))
n = tf.cast(tf.shape(a)[1], a.dtype)
b = tf.math.reduce_variance(a, axis=1, keepdims=True) * n / (n - 1)
print(b, b.shape)
```

```python
# 求最大值的第一种方法
a = tf.random.uniform([4, 3])
b = tf.reduce_max(a)
print(b)
```

```python
a = tf.random.uniform((4, 3))
values = tf.reduce_max(a, axis=1, keepdims=True)
indices = tf.argmax(a, axis=1)[:, None]
print(values, indices)
```

```python
# 求最小值的第一种方法
a = tf.random.uniform([4,3])
b = tf.reduce_min(a)
print(b)
```

```python
a = tf.random.uniform((4, 3))
values = tf.reduce_min(a, axis=1, keepdims=True)
indices = tf.argmin(a, axis=1)[:, None]
print(values, indices)
```

```python
a = tf.constant([1., 2., 3.])
b = tf.ones((3,))
output = tf.tensordot(a, b, axes=1)
print(output)
```

```python
# 向量逐元素相乘
a = tf.constant([1, 2, 3])
b = tf.constant([1, 1, 1])
output = tf.multiply(a, b)
print(output)
```

```python
a = tf.constant([1., 2., 3.])
b = tf.ones((3,))
output = tf.einsum("i,i->", a, b)
print(output)
```

```python
# 矩阵乘法
a = tf.constant([[1, 2, 3], [4, 5, 6]])
b = tf.constant([[1, 1], [2, 2], [3, 3]])
output = tf.matmul(a, b)
print(output)
```

```python
# 按批量相乘
a = tf.random.normal([10, 3, 4])
b = tf.random.normal([10, 4, 5])
output = tf.matmul(a, b)
print(output.shape)
# (10, 3, 5)
```

```python
# 用 Einstein 求和约定表达批量矩阵乘法。
print(tf.einsum('bij,bjk->bik', a, b).shape)
```

```python
a = tf.Variable(tf.ones((1, 2)))
b = a.numpy()
a.assign_add(tf.fill((1, 2), 2.))
print(a.numpy(), b)  # b 保留转换时的值。
b += 2
print(a.numpy(), b)
```

```python
a = np.ones((1, 2))
b = tf.convert_to_tensor(a)
a += 2
print(a, b)  # 修改数组不会更新已创建的张量。
```
